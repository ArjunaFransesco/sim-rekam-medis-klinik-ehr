### 1. Ingest Clinical EHR Dataset from SQLite
Connect to local SQLite database and inspect patient cohorts and vital signs.


```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

conn = sqlite3.connect('../database/clinical_ehr.db')
df_patients = pd.read_sql_query('SELECT * FROM patients', conn)
conn.close()
print(f'Total cohort records: {len(df_patients)}')
df_patients.head()
```

### 2. Clinical Feature Distribution & Acuity Correlation
Analyze distribution of Systolic Blood Pressure, Heart Rate, SpO2, and CRP across triage categories.


```python
print(df_patients.groupby('triage_category')[['systolic_bp', 'heart_rate', 'oxygen_sat', 'crp', 'admission_risk']].mean())
```
